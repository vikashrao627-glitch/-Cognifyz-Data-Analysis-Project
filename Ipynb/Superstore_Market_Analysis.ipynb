```python
# Clean Review Text
def clean_review(text):
    text = str(text).lower()
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
# Delivery Percentage by Price Range
delivery_percentage = (
    price_delivery
    .pivot(
        index=price_col,
        columns=delivery_col,
        values="Percentage"
    )
    .reset_index()
)

delivery_percentage
```

```python
# Table Booking Percentage
price_booking["Percentage"] = (
    price_booking.groupby(price_col)["Restaurant_Count"]
    .transform(lambda x: x / x.sum() * 100)
).round(2)

price_booking
```

```python
# Booking Percentage Table
booking_percentage = (
    price_booking
    .pivot(
        index=price_col,
        columns=booking_col,
        values="Percentage"
    )
    .reset_index()
)

booking_percentage
```

```python
# Price vs Table Booking Chart
booking_plot = price_booking.pivot(
    index=price_col,
    columns=booking_col,
    values="Restaurant_Count"
)

booking_plot.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Table Booking Availability by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Combined Service Analysis
service_analysis = (
    df.groupby(price_col)
    .agg(
        Restaurant_Count=(price_col, "size")
    )
    .reset_index()
)

service_analysis
```

```python
delivery_yes = (
    df[delivery_col].astype(str).str.strip().str.lower() == "yes"
)

delivery_by_price = (
    df.assign(Delivery_Yes=delivery_yes)
    .groupby(price_col)["Delivery_Yes"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index(name="Online_Delivery_Percentage")
)
```

```python
booking_yes = (
    df[booking_col].astype(str).str.strip().str.lower() == "yes"
)

booking_by_price = (
    df.assign(Booking_Yes=booking_yes)
    .groupby(price_col)["Booking_Yes"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index(name="Table_Booking_Percentage")
)
```

```python
# Combined Services Chart
service_analysis = service_analysis.merge(delivery_by_price, on=price_col, how='left')
service_analysis = service_analysis.merge(booking_by_price, on=price_col, how='left')

service_chart = service_analysis.set_index(price_col)[
    [
        "Online_Delivery_Percentage",
        "Table_Booking_Percentage"
    ]
]

service_chart.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Service Availability by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Percentage of Restaurants")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
